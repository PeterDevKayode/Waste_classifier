## 6. Compare the two models side by side

```python
import json, os

rows = []
for name in ["baseline_cnn_waste", "mobilenet_waste"]:
    path = f"models/{name}_metrics.json"
    if os.path.exists(path):
        with open(path) as f:
            rows.append(json.load(f))

import pandas as pd
pd.DataFrame(rows)
```
